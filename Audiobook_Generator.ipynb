```python
from TTS.api import TTS
import os
import re
from gtts.tokenizer import pre_processors, tokenizer_cases
from gtts.tokenizer.core import Tokenizer

# Initialize TTS with a specific model
tts = TTS(model_name="tts_models/en/jenny/jenny")

# Function to read lines from a file
def read_lines(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
    return lines

# Function to read the last processed line index
def read_last_index(index_file):
    if os.path.exists(index_file):
        with open(index_file, 'r') as file:
            return int(file.read().strip())
    return 0

# Function to write the last processed line index
def write_last_index(index_file, index):
    with open(index_file, 'w') as file:
        file.write(str(index))

# Pre-processing function to handle abbreviations
def preprocess_text(text):
    # Custom abbreviation fix
    text = re.sub(r'\b([A-Z]+)\b', lambda match: '-'.join(match.group(1)), text)
    # Default pre-processors
    text = pre_processors.tone_marks(text)
    text = pre_processors.end_of_line(text)
    text = pre_processors.abbreviations(text)
    text = pre_processors.word_sub(text)
    return text


# Function to generate audio files
def generate_audio_files(lines, lines_per_clip, output_dir, index_file):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    last_index = read_last_index(index_file)
    for i in range(last_index, len(lines), lines_per_clip):
        clip_text = ''.join(lines[i:i + lines_per_clip])
        # Pre-process and tokenize text
        clip_text = preprocess_text(clip_text)

        # Instead of tokenizing and then combining, we directly process the clip text
        output_path = os.path.join(output_dir, f'clip_{i//lines_per_clip + 1}.wav')
        print(f"Processing clip {i//lines_per_clip + 1}...")
        tts.tts_to_file(text=clip_text, file_path=output_path)
        print(f"Generated: {output_path}")

        # Update the last processed index
        write_last_index(index_file, i + lines_per_clip)


# User input for lines per clip
lines_per_clip = 10
# Ensure a text file has been uploaded
filename = '/content/drive/MyDrive/aws-whitepaper-short.txt'
index_file = '/content/drive/MyDrive/TTS-new/aws-whitepaper-short/last_index.txt'

# Read lines from the uploaded file
lines = read_lines(filename)

# Generate audio files
output_dir = "/content/drive/MyDrive/TTS-new/aws-whitepaper-short"
generate_audio_files(lines, lines_per_clip, output_dir, index_file)

print("Audio files have been generated in the 'TTS' directory.")
```
